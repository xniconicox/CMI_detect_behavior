````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../data/train.csv")

# クロス集計対象カテゴリ
cat_cols = ["subject", "sequence_id", "sequence_type", "orientation", "behavior", "phase"]

for col in cat_cols:
    ct = pd.crosstab(df[col], df["gesture"], normalize='index')
    if ct.shape[0] > 50:
        ct = ct.head(50)  # 上位50カテゴリで制限（表示しきれないとき）

    plt.figure(figsize=(12, min(15, 0.5 * ct.shape[0])))
    sns.heatmap(ct, cmap="viridis", annot=False)
    plt.title(f"{col} × gesture 分布（割合）")
    plt.xlabel("gesture")
    plt.ylabel(col)
    plt.show()
````

````python
# 例：ユニーク件数
for col in ["subject", "sequence_id", "sequence_type"]:
    print(f"{col}: unique={df[col].nunique():4d}  rows={len(df)}")
````

````python
train = df

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- ① pivot テーブルを作成（0/1）
pivot = pd.crosstab(train["subject"], train["gesture"])
binary = (pivot > 0).astype(int)   # 0 or 1

# --- ② ヒートマップ描画
plt.figure(figsize=(14, 10))
sns.heatmap(binary, cmap="Greys", cbar=False, vmin=0, vmax=1,
            linewidths=0.3, linecolor="lightgrey")
plt.title("Presence of each gesture per subject (1=present, 0=absent)")
plt.ylabel("subject")
plt.xlabel("gesture")
plt.xticks(rotation=70, ha="right")
plt.tight_layout()
plt.show()
````

````python
# 各 subject が何種類の gesture を持っているか
present_cnt = binary.sum(axis=1)
missing_cnt = binary.shape[1] - present_cnt
print("◆ gesture が欠けている subject 上位:")
print(missing_cnt.sort_values(ascending=False).head())
````

````python
# 各 gesture を持っていない subject 数
missing_per_gesture = (binary == 0).sum(axis=0)
print("◆ 多くの subject に欠けている gesture 上位:")
print(missing_per_gesture.sort_values(ascending=False).head())
````

````python
from sklearn.model_selection import GroupKFold
import pandas as pd
import numpy as np

N_FOLDS = 5
SEED = 42

def add_subject_folds(df, n_folds=N_FOLDS, seed=SEED):
    rng = np.random.RandomState(seed)
    subjects = df["subject"].unique()
    rng.shuffle(subjects)                    # ランダムに並べ替え

    # 均等に分割
    folds = np.zeros(len(df), dtype=int)
    for fold, subj_chunk in enumerate(np.array_split(subjects, n_folds)):
        folds[df["subject"].isin(subj_chunk)] = fold
    df = df.copy()
    df["fold"] = folds
    return df

train = add_subject_folds(train)

# --- 確認：Target / Non‑target 比率 ---
ratio = (train.groupby("fold")["sequence_type"]
                .value_counts(normalize=True)
                .unstack(fill_value=0))
print("Target / Non‑target 比率 (%)")
print((ratio * 100).round(1))

# ★ もし偏り >±5 % なら SEED を変えて再実行
````

### 📑 ここまでの確認結果と今後の方針まとめ

| 項目                         | 結論・決定事項                                    | 備考                                        |
| -------------------------- | ------------------------------------------ | ----------------------------------------- |
| **gesture × subject の網羅性** | **全 81 名すべてが 30 種の gesture を網羅**（欠落 0 件）   | 分割で gesture が欠ける心配なし                      |
| **分割単位**                   | **subject 単位の完全分離**で OK                    | `GroupKFold(subject)` を採用                 |
| **fold 数**                 | 5‑fold                                     | 1 fold ≒ 16 名                             |
| **Target / Non‑Target 比率** | 各 fold ≒ 60 : 40（±0.3 %）                   | 許容範囲内（±5 % 以内）                            |
| **リーク対策**                  | subject を跨いで学習/検証が混在しない                    | `sequence_id` も自動的に分離される                  |
| **希少クラス対策**                | 不要（網羅済み）                                   |                                           |
| **学習手順**                   | 各 fold でモデル初期化 → 学習 → `model_fold_k.h5` 保存 | IMU: Bi‑LSTM／ToF: 3D‑ResNet／Tabular: 残差付き |
| **推論手順**                   | 5 モデルをロード → softmax 出力を平均                  | エンドツーエンドでアンサンブル                           |
| **追加の前処理**                 | test に無い列は使用しない                            | （subject などカテゴリカルも除外）                     |
